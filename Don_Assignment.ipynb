**Valid/Invalid : Mobile/Email**

```python
def file_read(filename):
    fp = open(filename)
    data = fp.read()
    return data

def filter_mobile_num(filedata):
    valid_mobilenos = []
    invalid_mobilenos= []
    for word in filedata.split():
        if word.isnumeric():
            if len(word)==10:
                valid_mobilenos.append(word)
            else:
                invalid_mobilenos.append(word)
    return valid_mobilenos,invalid_mobilenos
            

def filter_email_ids(filedata):
    valid_email = []
    invalid_email = []

    for word in filedata.split():
        if word.endswith('@gmail.com'):
            username = word[:-len('@gmail.com')]
            if username.isalnum():
                valid_email.append(word)
            else:
                invalid_email.append(word)
    return valid_email, invalid_email

def main():
    filename = 'valid_or_not.txt'
    fdata = file_read(filename)
    v_mobile,in_mobile = filter_mobile_num(fdata)
    print("\nValid mobile number\t",v_mobile)
    print("\nInvalid mobile number\t",in_mobile)
    v_email, inv_email = filter_email_ids(fdata)
    print("\nValid  email\t",v_email)
    print("\nInvalid email\t",inv_email)

if __name__=='__main__':
    main()
```

```python
def file_read(filename):
    with open(filename) as fp:
        return fp.read()


def count_letters(filedata):
    count = 0
    for ch in filedata:
        if ch.isalpha():
            count += 1
    return count


def count_words(filedata):
    return len(filedata.split())


def count_lines(filedata):
    return len(filedata.splitlines())


def count_special_chars(filedata):
    sp = []
    for ch in filedata:
        if not ch.isalnum() and not ch.isspace():   # not letter, not digit, not space
            sp.append(ch)
    return len(sp), sp


def count_numeric(filedata):
    numeric_words = []
    for word in filedata.split():
        if word.isnumeric():
            numeric_words.append(word)
    return len(numeric_words), numeric_words


def main():
    filename = 'sample.txt'
    fdata = file_read(filename)

    total_letters = count_letters(fdata)
    total_chars = len(fdata)                 # every character, including spaces and newlines
    total_words = count_words(fdata)
    total_lines = count_lines(fdata)
    total_special, special_list = count_special_chars(fdata)
    total_numeric, numeric_list = count_numeric(fdata)

    print("Total Characters\t:", total_chars)
    print("Total Letters\t\t:", total_letters)
    print("Total Words\t\t:", total_words)
    print("Total Lines\t\t:", total_lines)
    print("Total Special Chars\t:", total_special, "\t", special_list)
    print("Total Numeric Values\t:", total_numeric, "\t", numeric_list)


if __name__ == '__main__':
    main()
```

**Create 3 sepearte files i.e chars, special chars, numeric**

```python
def file_read(filename):
    with open(filename, encoding='utf-8') as fp:
        return fp.read()

def file_write(filename, data):
    with open(filename, 'w', encoding='utf-8') as fp:
        fp.write(data)
    
def get_chars(filename):
    ch = []
    for chars in filename:
        if chars.isalpha():
            ch.append(chars)
    return '\n'.join(ch)

def spe_chars(filename):
    spc = []
    for ch in filename:
        if not ch.isalpha() and not ch.isspace() and not ch.isnumeric():
            spc.append(ch)
    return '\n'.join(spc)

def nume(filename):
    nw = []
    for word in filename.split():
        if word.isnumeric():
            if word.isnumeric():
                nw.append(word)
    return '\n'.join(nw)

def main():
    sourcefile = 'output.txt'
    fdata = file_read(sourcefile)

    char_data = get_chars(fdata)
    special_chars = spe_chars(fdata)
    numeric_data = nume(fdata)

    file_write('chars.txt',char_data)
    print("\nnew char file created")
    file_write('special_chars.txt', special_chars)
    print("\nnew special char file created")
    file_write('numeric.txt', numeric_data)
    print("\nnew numeric file created")


if __name__ == "__main__":
    main()
```
